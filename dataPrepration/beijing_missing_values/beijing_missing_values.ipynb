<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
پیش از این که به مدیریت داده‌های گم‌شده بپردازیم٬ بهتر است که کمی با مجموعه‌داده‌ی خانه‌های شهر پکن آشنا شویم. در درسنامه‌‌ی <a href="https://quera.org/college/9516/chapter/33703/lesson/114220/?comments_page=1&comments_filter=ALL"><b>"معرفی مجموعه‌داده درسنامه"</b></a> در فصل مقدمه٬ اطلاعات و توضیحات کامل راجع به ویژگی‌های این مجموعه‌داده جمع‌آوری شده است. اگر با این مجموعه‌داده آشنایی ندارید٬ حتما این درسنامه را مطالعه کنید.
<br>
ابتدا نیاز است فایل مجموعه‌داده (<code>beijing.csv</code>) را بخوانیم.
<br>
 اگر کمی مجموعه‌داده را بررسی کرده باشید متوجه خواهید شد که در بعضی از ستون‌ها عبارتی مثل <code>未知</code> و <code>unknown</code> وجود دارند. به کمک مترجم گوگل می‌توان متوجه شد که <code>未知</code> نیز به معنی <code>unknown</code> یعنی ناشناخته است. بنابراین در هنگام خواندن مجموعه‌داده می‌توانیم به کمک آرگومان <code>na_values</code> این دو عبارت را نیز معادل <code>NaN</code> لحاظ کنیم.
<br>
علاوه بر این از آنجا که ستون <code>tradeTime</code> باید از نوع تاریخ‌زمان (<code>datetime</code>) باشد، به کمک آرگومان <code>parse_dates</code> از همان ابتدا این تبدیل نوع را هم انجام می‌دهیم. 
</p>

```python
import pandas as pd
# reading the dataset and replacing the unknown values with NaN
df = pd.read_csv('beijing.csv', encoding='gbk', low_memory=False, parse_dates=['tradeTime'], na_values=['unknown','未知'])
```

<h2 dir="rtl" align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
مدیریت مقادیر گم‌شده
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
یکی از روش‌هایی که در این مجموعه‌داده برای ما کاربردی ندارد حذف ستون است ولی بهتر است که با آن آشنا شویم. نحوه‌ی اجرای این روش این است که تعداد داده‌های گم‌شده در یک ستون را بررسی کرده و اگر این تعداد بیشتر از یک حد آستانه (به‌طور مثال ۹۰ درصد) باشد، می‌توان آن ستون را به‌کلی حذف کرد. زیرا وجود آن ستون به دلیل مقادیر زیادی داده‌ی گم‌شده به احتمال بالا اطلاعات خاصی را برای‌مان فراهم نمی‌کند.
</font>
</p>

```python
missing_percentages = df.isnull().mean()
columns_to_delete = missing_percentages[missing_percentages > .9].index
print(columns_to_delete)
df.drop(columns = columns_to_delete, inplace = True)
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
    <font face="vazirmatn" size=3>
مشاهده می‌کنیم در هیچکدام از ستون‌ها مقادیر گم‌شده بیشتر از ۹۰ درصد نیست و نیازی به حذف نداریم.
برای مدیریت باقی موارد در هر ستون، روش خاصی انتخاب کرده‌ایم که در ادامه آن‌ها را با هم بررسی می‌کنیم.

</font>
</p>

</font>
</p>

<h3 dir="rtl" align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
ستون <code>communityAverage</code>
</font>
</h3>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
    با اجرای سلول زیر به عنوان مثال متوجه می‌شویم ۳۵۸۶تا <code>Cid</code> وجود دارد که همه‌ی خانه‌هایشان تنها یک قیمت میانگین داشته‌اند و فقط ۷تا <code>Cid</code> وجود دارد که برای خانه‌هایش سه قیمت میانگین ثبت شده است.
</font>
</p>

```python
df.groupby('Cid')['communityAverage'].nunique().value_counts()
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
برای پر کردن مقادیر گم‌شده‌ی این ستون بهترین کار این است که میانگین واقعی را  به تفکیک <code>cid</code> <b>از روی ستون <code>price</code> محاسبه کنیم</b> و مقادیر خالی را جایگزین نمایید. به بیانی دیگر٬ بهتر است برای ردیف‌هایی که مقدار این ستون را از دست داده‌اند، مقدار <code>communityAverage</code>  را با میانگین <code>price</code> خانه‌های دیگر موجود در آن مجتمع (یعنی خانه‌هایی که <code>Cid</code> آن‌ها با مقدار <code>Cid</code> ردیفی که مقدار <code>communityAverage</code>  را از دست داده است برابر باشند) پر کنیم.
</font>
</p>

```python
mean = df.groupby('Cid')['price'].transform('mean')
df.loc[:, 'communityAverage'] = df['communityAverage'].fillna(mean)
```

```python
df['communityAverage'].isna().sum()
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
با اجرای سلول بالا، تعداد مقادیر گم‌شده از ۴۶۳ به ۰ کاهش پیدا می‌کند. شاید برای ما سوالی پیش بیاید٬ که چرا از میانگین خود ستون <code>communityAverage</code> استفاده نکردیم. زیرا ستون <code>communityAverage</code>ممکن است <b>قدیمی یا  به روش متفاوتی محاسبه شده<b> باشد و </b>(محاسبه میانگین واقعی هر <code>Cid</code>)</b> روش بهتری است، زیرا داده‌ها را بر اساس واقعیت جامعه به‌روز می‌کند. همچنین اگر بررسی کنیم٬ می‌بینیم که این روش به یکپارچگی مقادیر این ستون آسیبی نرسانده است.
</font>
</p>

```python
df.groupby('Cid')['communityAverage'].nunique().value_counts()
```

```python
df.isna().sum()
```

<h3 dir="rtl" align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazirmatn" color="#0099cc">
ستون <code>buildingStructure</code>
</font>
</h3>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
با مشاهده تعداد تکرار موارد منحصربه‌فرد موجود در این ستون متوجه خواهیم شد که ساختار ساختمانی حدود ۵۹ درصد از سطرها <code>steel-concrete composite</code> می‌باشد. بنابراین مقادیر گم‌شده سایر ردیف‌ها را نیز می‌توانیم با همین مقدار اکثریت پر کنیم. همچنین، برای آن‌که بدانیم برخی از ردیف‌های این ستون، دارای مقادیر گم‌شده بودند، ستون دیگری به اسم <code>buildingStructure_ismissing</code> به مجموعه‌داده اضافه می‌کنیم.
</font>
</p>

```python
df['buildingStructure'].unique()
```

```python
df['buildingStructure'].value_counts(normalize=True)
```

```python
df['buildingStructure_ismissing'] = df['buildingStructure'].isna()
df['buildingStructure'].fillna('steel-concrete composite', inplace=True)
```

```python
df.isna().sum()
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
در هر ردیفی که مقدار ستون <code>buildingType</code> آن را از دست داده، بیشترین نوعی را جایگزین می‌کنیم که در منطقه‌ی آن خانه خاص، وجود داشته است.
</font>
</p>

```python
df['buildingType'].value_counts(normalize=True)
```

```python
df['buildingType_ismissing'] = df['buildingType'].isna()
frequent_types = df.groupby('district')['buildingType'].transform(lambda x: x.mode()[0])
df['buildingType'].fillna(frequent_types, inplace=True)
```

```python
df.isna().sum()
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
با رسم نمودار جعبه‌ای از ستون <code>constructionTime</code> متوجه وجود مقادیر پرت در این ستون می‌شویم.

```python
import plotly.express as px
fig = px.box(df, y='constructionTime')
fig.update_traces(marker_color='#a83277')
fig.update_layout(autosize=False, width=600, height=500, title='Detected Outliers')
fig.show()
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
به کمک سلول زیر می‌توانیم نقشه‌ی پراکندگی این ستون را رسم کنیم:
</font>
</p>

```python
%matplotlib inline
from matplotlib import pyplot as plt

fig , ax = plt.subplots()
fig.set_size_inches(12.5, 7)

df.plot(x='Lng', y='Lat', ax=ax, kind='scatter', alpha=0.4, c='constructionTime', cmap=plt.get_cmap('jet'), colorbar=True)
ax.axis('equal')
beijing_map = plt.imread('map1.jpg')
ax.imshow(beijing_map, extent=[115.89777890444654, 116.90711309555346, 39.5957436, 40.2840444])
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
این‌بار مقادیر پرت را از مجموعه‌داده حذف کرده و مجدداً نقشه‌ی پراکندگی را رسم می‌کنیم.
</font>
</p>

```python
Q1 = df['constructionTime'].quantile(0.25)
Q3 = df['constructionTime'].quantile(0.75)
fence = 1.5 * (Q3 - Q1)
plot_df = df[(df['constructionTime'] > Q1 - fence) & (df['constructionTime'] < Q3 + fence)]

fig , ax = plt.subplots()
fig.set_size_inches(12.5, 7)

plot_df.plot(x='Lng', y='Lat', ax=ax, kind='scatter', alpha=0.4, c='constructionTime', cmap=plt.get_cmap("jet"), colorbar=True)
ax.axis('equal')
beijing_map = plt.imread('map1.jpg')
ax.imshow(beijing_map, extent=[115.89777890444654, 116.90711309555346, 39.5957436, 40.2840444])
```

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazirmatn" size=3>
برای پر کردن مقادیر گم‌شده‌ی این ستون، به تفکیک منطقه، سالی که بیشترین خانه در آن ساخته شده است را انتخاب می‌کنیم.
</font>
</p>

```python
years = df.groupby('district')['constructionTime'].transform(lambda x : x.mode()[0])
df['constructionTime'].fillna(years, inplace=True)
```

```python
df.isna().sum()
```
